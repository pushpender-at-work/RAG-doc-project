#### document loader in langchian 
Document Loaders

Definition: A document loader is a component that reads data from a source and converts it into a common document format that an LLM/RAG pipeline can process.

Key idea:
Source → Document Loader → Documents

Common loaders:

PDF → PyPDFLoader

TXT → TextLoader

CSV → CSVLoader

Web pages → WebBaseLoader

DOCX → Docx2txtLoader

```python
# text data 
from langchain_community.document_loaders import TextLoader
loader=TextLoader('text.txt')

text_doc=loader.load()
text_doc

```

```python
#pdfloader
from langchain_community.document_loaders import PyPDFLoader
loader=PyPDFLoader("pdf_name.pdf")
pdf_doc=loader.load()

```

#### Document Splitters

Definition: A document splitter takes large documents and breaks them into smaller chunks so they fit efficiently into the LLM's context window and can be embedded/retrieved effectively.

```python
#  recurcive text splitter 
# used to recurcively split text into chuncks and make socuments from them 
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = splitter.split_documents("documents")
```
