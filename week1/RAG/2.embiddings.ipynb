### Embeddings

Convert each text chunk into a numerical vector that represents its semantic meaning.

Chunk → Embedding Model → Vector

#### Vector Database / Vector Store

Stores the embeddings and allows you to search for semantically similar chunks.

Chunks → Embeddings → Vector DB

Examples: FAISS, Chroma, Pinecone, Qdrant, Weaviate.



#### Similarity Search / Retrieval

When the user asks a question, convert the question into an embedding and retrieve the most relevant chunks.

Question

   ↓

Query Embedding

   ↓

Similarity Search

   ↓

Relevant Chunks





Embedding ≠ similarity search.

Embedding: converts text → vector.
Similarity search: compares vectors to find relevant content.

Vector database: stores and searches those vectors.


##### vector database implementation using hugging face

```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.document_loaders import TextLoader

loader=TextLoader("text.txt")
Text_doc=loader.load()


splitter = RecursiveCharacterTextSplitter(
    chunk_size=50,
    chunk_overlap=10
)

docs=splitter.split_documents(Text_doc)


embidding=HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

vector_db = Chroma.from_documents(docs, embidding)

retriver=vector_db.similarity_search(" what is fast api ")


```

##### FAISS

Definition: FAISS (Facebook AI Similarity Search) is a library for storing, indexing, and searching vectors efficiently. In a RAG system, it is commonly used to find document chunks whose embeddings are most similar to a user's query embedding.

Documents
   ↓
Chunks
   ↓
Embedding Model
   ↓
Vectors
   ↓
FAISS Index
   ↓
Similarity Search
   ↓
Top-K relevant chunks

```python
import faiss
import numpy as np
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

documents = [
    "FastAPI is a Python web framework.",
    "RAG retrieves relevant documents before generating an answer.",
    "Embeddings convert text into numerical vectors."
]

embeddings = model.encode(documents)

# FAISS expects float32
embeddings = np.array(embeddings).astype("float32")

dimension = embeddings.shape[1]

index = faiss.IndexFlatL2(dimension)
index.add(embeddings)

query = "How does RAG work?"
query_embedding = model.encode([query]).astype("float32")

distances, indices = index.search(query_embedding, k=2)

print(indices)
print(distances)
```
