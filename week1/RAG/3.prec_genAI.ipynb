#### Getting started With Langchain And Open AI

In this quickstart we'll see how to:

- Get setup with LangChain, LangSmith and LangServe
- Use the most basic and common components of LangChain: prompt templates, models, and output parsers.
- Build a simple application with LangChain
- Trace your application with LangSmith
- Serve your application with LangServe

```python
import os 
from dotenv import load_dotenv
load_dotenv()

GOOGLE_API_KEY=os.getenv("GEMINI_API_KEY")
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm=ChatGoogleGenerativeAI(
    model='gemini-3.5-flash',
    temperature=0.7)


```

```python
## input and get response from llm 
response=llm.invoke(" what is generative ai ")
```

```python
## prompt tempelate ( descibe how the llm mpdel wil achtually work )
# chat pormpt tempelate 
from langchain_core.prompts import ChatPromptTemplate

prompt=ChatPromptTemplate.from_messages(
    [
        ('system','you are an expert AI enginner . Provide me answers based on the qustions. give me short answers '),
        ('user', '{input}')
    ]
)

```

```python
# chain ( this is used to combine multiple porcudurs to run along side )
chain=prompt|llm

response=chain.invoke({'input':"can you tell me about langsmith"})
print(response)
```

```python
# now we can not take only string output from this so we sill use 
# output parsers 
from langchain_core.output_parsers import StrOutputParser
output_parser=StrOutputParser()
chain=prompt|llm|output_parser

response=chain.invoke({'input':" tel me about langsmith "})
```
